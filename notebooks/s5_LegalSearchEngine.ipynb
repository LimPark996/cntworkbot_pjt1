# 🔍 하이브리드 검색 엔진 만들기

## 🎯 실습 목표
- **벡터 검색**과 **키워드 검색**을 결합한 하이브리드 검색 구현
- BM25 알고리즘으로 키워드 기반 검색 이해하기
- RRF(Reciprocal Rank Fusion)로 검색 결과 융합하기
- 법령 특화 필터링 (조/장 단위) 적용하기

---

## 목적
RAG 시스템에서 사용자 질문에 가장 관련성 높은 문서를 찾기 위한 검색 엔진입니다. 벡터 검색만으로는 키워드 매칭이 약하고, 키워드 검색만으로는 의미적 유사성을 놓칠 수 있습니다. 두 방식을 결합한 하이브리드 검색으로 검색 품질을 높입니다.

## 기능
FAISS를 이용한 벡터 유사도 검색과 BM25를 이용한 키워드 검색을 동시에 수행합니다. 두 검색 결과를 RRF 알고리즘으로 융합하여 최종 순위를 결정합니다. 법령 문서 특성에 맞게 "제36조", "제2장" 등 조/장 단위 필터링을 지원합니다.

## 결과물
검색 쿼리에 대해 관련성 높은 청크들을 순위별로 반환합니다. 각 결과에는 청크 내용, 메타데이터, 검색 점수가 포함됩니다. 이 검색 결과는 다음 단계인 LLM 답변 생성의 컨텍스트로 사용됩니다.

## 🔧 Step 2: 임베딩 함수 (s4에서 가져오기)

검색 시 쿼리를 임베딩해야 하므로 s4의 함수들을 사용합니다.

````python
# 임베딩 관련 함수들 (s4에서 복사)
embedding_cache = {}

def get_text_hash(text: str) -> str:
    """텍스트의 MD5 해시 계산"""
    return hashlib.md5(text.encode('utf-8')).hexdigest()

def embed_text(text: str, 
               client: OpenAI,
               model: str = "text-embedding-3-large",
               dimension: int = 3072) -> np.ndarray:
    """텍스트를 임베딩 벡터로 변환"""
    global embedding_cache
    
    text_hash = get_text_hash(text)
    if text_hash in embedding_cache:
        return embedding_cache[text_hash]
    
    try:
        response = client.embeddings.create(
            input=text,
            model=model
        )
        embedding = np.array(response.data[0].embedding, dtype='float32')
        embedding_cache[text_hash] = embedding
        return embedding
    except Exception as e:
        print(f"⚠ 임베딩 생성 실패: {e}")
        return np.zeros(dimension, dtype='float32')

print("✅ 임베딩 함수 정의 완료")
````

## 🔧 Step 3: 인덱스 및 메타데이터 로드

````python
def load_index(index_path: str) -> Optional[faiss.Index]:
    """FAISS 인덱스 로드"""
    if not os.path.exists(index_path):
        print(f"⚠ 인덱스 없음: {index_path}")
        return None
    
    index = faiss.read_index(index_path)
    print(f"✓ 인덱스 로드: {index.ntotal}개 벡터")
    return index

def load_metadata(metadata_path: str) -> Optional[List[Dict]]:
    """메타데이터 로드"""
    if not os.path.exists(metadata_path):
        print(f"⚠ 메타데이터 없음: {metadata_path}")
        return None
    
    with open(metadata_path, 'r', encoding='utf-8') as f:
        metadata = json.load(f)
    print(f"✓ 메타데이터 로드: {len(metadata)}개 항목")
    return metadata

print("✅ 로드 함수 정의 완료")
````

## 🔧 Step 6: 법령 필터링

법령 문서는 "제36조", "제2장" 등 특정 조/장으로 필터링이 필요합니다.

````python
def contains_article(content: str, article: str) -> bool:
    """
    청크 내용에 특정 조(條)가 포함되어 있는지 확인
    
    Args:
        content: 청크 내용
        article: 조 번호 (예: "제36조")
    
    Returns:
        포함 여부
    """
    pattern = re.escape(article) + r'(?:\s|[^\w가-힣]|$)'
    return bool(re.search(pattern, content))

def contains_chapter(content: str, chapter: str) -> bool:
    """청크 내용에 특정 장(章)이 포함되어 있는지 확인"""
    pattern = re.escape(chapter) + r'(?:\s|[^\w가-힣]|$)'
    return bool(re.search(pattern, content))

print("✅ 필터링 함수 정의 완료")
````

## 🔧 Step 7: 키워드 검색 (BM25)

````python
def keyword_search(query: str,
                   bm25: BM25Okapi,
                   metadata: List[Dict],
                   top_k: int = 5) -> List[Dict]:
    query_tokens = tokenize_korean(query)
    scores = bm25.get_scores(query_tokens)
    ranked_indices = np.argsort(scores)[::-1]
    
    results = []
    for idx in ranked_indices:
        if scores[idx] <= 0:
            continue
        
        item = metadata[idx]
        result = {
            "rank": len(results) + 1,
            "chunk_id": item["chunk_id"],
            "content": item["content"],
            "score": float(scores[idx])
        }
        results.append(result)
        
        if len(results) >= top_k:
            break
    
    return results
````

## 🔧 Step 8: 벡터 검색 (FAISS)

````python
def vector_search(query: str,
                  faiss_index: faiss.Index,
                  metadata: List[Dict],
                  client: OpenAI,
                  top_k: int = 5) -> List[Dict]:
    """
    FAISS 벡터 검색을 수행합니다.
    
    Args:
        query: 검색 쿼리
        faiss_index: FAISS 인덱스
        metadata: 메타데이터
        client: OpenAI 클라이언트
        top_k: 반환할 결과 수
    
    Returns:
        List[Dict]: 검색 결과
    """
    # 쿼리 임베딩
    query_embedding = embed_text(query, client)
    query_vector = query_embedding.reshape(1, -1).astype('float32')
    
    # FAISS 검색
    distances, indices = faiss_index.search(query_vector, top_k)
    
    # 결과 구성
    results = []
    for i, (idx, distance) in enumerate(zip(indices[0], distances[0])):
        if idx >= len(metadata):
            continue
        
        item = metadata[idx]
        result = {
            "rank": i + 1,
            "chunk_id": item["chunk_id"],
            "content": item["content"],
            "metadata": item.get("metadata", {}),
            "score": float(1 / (1 + distance)),  # 거리를 유사도로 변환
            "search_type": "vector"
        }
        results.append(result)
    
    return results

print("✅ 벡터 검색 함수 정의 완료")
````

## 🔧 Step 9: RRF (Reciprocal Rank Fusion)

### RRF란?
- 여러 검색 결과를 **융합**하는 알고리즘
- 각 결과의 순위(rank)를 기반으로 점수 계산
- 공식: `score = 1 / (k + rank)`

````python
def reciprocal_rank_fusion(vector_results: List[Dict],
                           keyword_results: List[Dict],
                           k: int = 60) -> List[Dict]:
    chunk_scores = {}
    chunk_data = {}
    
    # 벡터 검색 결과 처리
    for result in vector_results:
        chunk_id = result["chunk_id"]
        rank = result["rank"]
        rrf_score = 1 / (k + rank)
        
        chunk_scores[chunk_id] = chunk_scores.get(chunk_id, 0) + rrf_score
        chunk_data[chunk_id] = result
    
    # 키워드 검색 결과 처리
    for result in keyword_results:
        chunk_id = result["chunk_id"]
        rank = result["rank"]
        rrf_score = 1 / (k + rank)
        
        chunk_scores[chunk_id] = chunk_scores.get(chunk_id, 0) + rrf_score
        if chunk_id not in chunk_data:
            chunk_data[chunk_id] = result
    
    # 점수순 정렬
    sorted_chunks = sorted(chunk_scores.items(), key=lambda x: x[1], reverse=True)
    
    # 결과 구성
    results = []
    for i, (chunk_id, score) in enumerate(sorted_chunks):
        result = chunk_data[chunk_id].copy()
        result["rank"] = i + 1
        result["rrf_score"] = float(score)
        result["search_type"] = "hybrid"
        results.append(result)
    
    return results
````

### ✅ 테스트 #4: RRF 융합

````python
# RRF 테스트용 더미 데이터
vector_results = [
    {"rank": 1, "chunk_id": "chunk_002", "content": "건폐율...", "score": 0.95},
    {"rank": 2, "chunk_id": "chunk_003", "content": "용적률...", "score": 0.85},
    {"rank": 3, "chunk_id": "chunk_001", "content": "건축법...", "score": 0.75},
]

keyword_results = [
    {"rank": 1, "chunk_id": "chunk_002", "content": "건폐율...", "score": 5.2},
    {"rank": 2, "chunk_id": "chunk_001", "content": "건축법...", "score": 3.1},
    {"rank": 3, "chunk_id": "chunk_004", "content": "건축허가...", "score": 2.5},
]

fused_results = reciprocal_rank_fusion(vector_results, keyword_results)

print(f"✅ RRF 융합 완료")
print(f"\n융합 결과: {len(fused_results)}건")
for r in fused_results:
    print(f"  [{r['rank']}] {r['chunk_id']} - RRF 점수: {r.get('rrf_score', 0):.4f}")
````

## 🔧 Step 10: 하이브리드 검색

````python
def hybrid_search(query: str,
                  faiss_index: faiss.Index,
                  bm25: BM25Okapi,
                  metadata: List[Dict],
                  client: OpenAI,
                  top_k: int = 5,
                  filter_article: Optional[str] = None,
                  filter_chapter: Optional[str] = None) -> List[Dict]:
    """
    하이브리드 검색 (벡터 + 키워드 + RRF)
    
    Args:
        query: 검색 쿼리
        faiss_index: FAISS 인덱스
        bm25: BM25 모델
        metadata: 메타데이터
        client: OpenAI 클라이언트
        top_k: 반환할 결과 수
        filter_article: 조 필터 (예: "제36조")
        filter_chapter: 장 필터 (예: "제2장")
    
    Returns:
        List[Dict]: 검색 결과
    """
    # 벡터 검색
    if faiss_index is not None:
        v_results = vector_search(query, faiss_index, metadata, client, top_k=top_k*2)
    else:
        v_results = []
    
    # 키워드 검색
    k_results = keyword_search(query, bm25, metadata, top_k=top_k*2)
    
    # 필터링 적용
    if filter_article:
        v_results = [r for r in v_results if contains_article(r["content"], filter_article)]
        k_results = [r for r in k_results if contains_article(r["content"], filter_article)]
    
    if filter_chapter:
        v_results = [r for r in v_results if contains_chapter(r["content"], filter_chapter)]
        k_results = [r for r in k_results if contains_chapter(r["content"], filter_chapter)]
    
    # 순위 재조정
    for i, r in enumerate(v_results):
        r["rank"] = i + 1
    for i, r in enumerate(k_results):
        r["rank"] = i + 1
    
    # RRF 융합
    hybrid_results = reciprocal_rank_fusion(v_results, k_results)
    
    return hybrid_results[:top_k]

print("✅ 하이브리드 검색 함수 정의 완료")
````
